```python
import requests
import json
```

```python
# Base URL for the local Flask API
BASE_URL = 'http://localhost:5000/employees'
```

```python
# New employee data to add
new_employee = {
    'name': 'Liam'
}
```

```python
json.dumps(new_employee)
```

```python
# POST request to add the new employee
response = requests.post(BASE_URL, data=json.dumps(new_employee))
# Check if the request was successful
if response.status_code == 201:
    print("New employee added successfully.")
else:
    print(f"Failed to add employee. Status code: {response.status_code}, Response: {response.text}")
```

```python
# GET request to fetch all employees
response = requests.get(BASE_URL)
# Check and display the employee list
if response.status_code == 200:
    employees = response.json()
    print("Current list of employees:")
    for emp in employees:
        print(f"ID: {emp['id']}, Name: {emp['name']}")
else:
    print(f"Could not retrieve employee list. Status code: {response.status_code}, Response: {response.text}")
```
